# 05 — Full Pipeline 整链

整链组装 → render_movie → 导出多帧 TIFF。

```python
import numpy as np
from simulation_smile import SimulationPipeline, configs as C
from simulation_smile.optics import PupilPSF
from simulation_smile.illumination import UniformField
from simulation_smile.fluorophore import BlinkingModel
from simulation_smile.camera import SCMOSCamera
from simulation_smile.noise import Background, Poisson, ReadNoise
from simulation_smile.io import save_tiff

grid = C.Grid(pixel_size=108e-9, size=64, oversampling=4)
pipe = (SimulationPipeline(grid=grid)
        .with_optics(PupilPSF(C.PSFConfig(na=1.49), grid=grid))
        .with_illumination(UniformField(C.IlluminationConfig()))
        .with_photophysics(BlinkingModel(C.FluorophoreConfig()))
        .with_camera(SCMOSCamera(C.CameraConfig()))
        .with_noise([Background(C.NoiseConfig()), Poisson(), ReadNoise(C.NoiseConfig())]))

# movie = pipe.render_movie(200)
# save_tiff('out.tif', movie)
```
